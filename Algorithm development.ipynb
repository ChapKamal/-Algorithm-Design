##### A Simple Algorithm Design

```python
critical_materials = [
    "Lithium",    "Cobalt",     "Nickel",      "Manganese",  "Graphite",
    "Neodymium",  "Dysprosium", "Terbium",     "Cerium",     "Lanthanum",
    "Gallium",    "Germanium",  "Indium",      "Tellurium",  "Titanium",
    "Tungsten",   "Rhenium",    "Beryllium",   "Hafnium",    "Platinum",
    "Palladium",  "Rhodium",    "Iridium",     "Ruthenium",  "Chromium",
    "Magnesium",  "Antimony",   "Fluorspar",   "Vanadium",   "Scandium",
    "Yttrium",    "Niobium",    "Tantalum",    "Selenium",   "Silicon",
    "Boron",      "Bismuth",    "Tin",         "Zinc",       "Helium"
]
def search(x, materials_list):
    for materials in materials_list:
        if materials == x and materials in ["Lithium", "Cobalt", "Nickel", "Manganese", "Graphite"]:
            return "Materials are used in Energy & Clean Technology: " + x
        elif materials == x and materials in ["Neodymium", "Dysprosium", "Terbium", "Cerium", "Lanthanum"]:
            return "Materials are Rare Earth Elements which are used in Hightec & Electronics: " + x
        elif materials == x and materials in ["Gallium", "Germanium",  "Indium", "Tellurium"]:
            return "Materials are used in Hightec & Electronics: " + x
        elif  materials == x and materials in [ "Titanium","Tungsten", "Rhenium","Beryllium","Hafnium"]:
            return "Materials are used in Defense & Aerospace: " + x
        elif  materials == x and materials in ["Platinum","Palladium", "Rhodium","Iridium","Ruthenium", "Chromium",
                               "Magnesium", "Antimony",   "Fluorspar"]:
            return "Materials are used in Industry & speciality: " + x

    return "It is not a critical material: " + x
```

```python
print(search("Lithium", critical_materials))
print(search("Neodymium", critical_materials))
print(search("Titanium", critical_materials))
print(search("Gold", critical_materials))
```
